```python
#Data reads not fully accurate. Need to normalize tracts

def normalize_tract6(val):
    if pd.isna(val):
        return None
    val = val.strip().zfill(4)  
    return val.ljust(6, "0")    
```

```python
Pluto_2009["Tract2000"] = Pluto_2009["Tract2000"].apply(normalize_tract6)
Pluto_2010["Tract2000"] = Pluto_2010["Tract2000"].apply(normalize_tract6)
Pluto_2011["Tract2010"] = Pluto_2011["Tract2010"].apply(normalize_tract6)
Pluto_2012["Tract2010"] = Pluto_2012["Tract2010"].apply(normalize_tract6)
Pluto_2013["Tract2010"] = Pluto_2013["Tract2010"].apply(normalize_tract6)
Pluto_2014["Tract2010"] = Pluto_2014["Tract2010"].apply(normalize_tract6)
Pluto_2015["Tract2010"] = Pluto_2015["Tract2010"].apply(normalize_tract6)
Pluto_2016["Tract2010"] = Pluto_2016["Tract2010"].apply(normalize_tract6)
Pluto_2017["Tract2010"] = Pluto_2017["Tract2010"].apply(normalize_tract6)
Pluto_2018["tract2010"] = Pluto_2018["tract2010"].apply(normalize_tract6)
Pluto_2019["tract2010"] = Pluto_2019["tract2010"].apply(normalize_tract6)
Pluto_2020["tract2010"] = Pluto_2020["tract2010"].apply(normalize_tract6)
```

```python
def full_tract(df, tract_col, borough_num):

    df['full_tract'] = np.where(
        df[tract_col].notna(),
        "36" + borough_num + df[tract_col].astype(str).str.zfill(6),
        np.nan
    )

    return df
```

```python
Pluto_2009=full_tract(Pluto_2009, 'Tract2000', '061')
Pluto_2010=full_tract(Pluto_2010, 'Tract2000', '061')
Pluto_2011=full_tract(Pluto_2011, 'Tract2010', '061')
Pluto_2012=full_tract(Pluto_2012, 'Tract2010', '061')
Pluto_2013=full_tract(Pluto_2013, 'Tract2010', '061')
Pluto_2014=full_tract(Pluto_2014, 'Tract2010', '061')
Pluto_2015=full_tract(Pluto_2015, 'Tract2010', '061')
Pluto_2016=full_tract(Pluto_2016, 'Tract2010', '061')
Pluto_2017=full_tract(Pluto_2017, 'Tract2010', '061')
Pluto_2018_MN=full_tract(Pluto_2018_MN, 'tract2010', '061')
Pluto_2019_MN=full_tract(Pluto_2019_MN, 'tract2010', '061')
Pluto_2020_MN=full_tract(Pluto_2020_MN, 'tract2010', '061')
```

Aggregate and Concat Dataframes

```python
Pluto_2009['year']=2009
Pluto_2010['year']=2010
Pluto_2011['year']=2011
Pluto_2012['year']=2012
Pluto_2013['year']=2013
Pluto_2014['year']=2014
Pluto_2015['year']=2015
Pluto_2016['year']=2016
Pluto_2017['year']=2017
Pluto_2018_MN['year']=2018
Pluto_2019_MN['year']=2019
Pluto_2020_MN['year']=2020
```

```python
def Pluto_agg(df):
    df_agg = (
        df
        .groupby(['full_tract', 'year'])[
            ['LotArea', 'BldgArea', 'ComArea', 'ResArea', 'OfficeArea',
             'RetailArea', 'GarageArea', 'StrgeArea', 'FactryArea',
             'OtherArea', 'NumBldgs', 'NumFloors', 'UnitsRes', 'UnitsTotal', 
             'AssessLand', 'AssessTot']
        ]
        .agg('sum')
        .reset_index()
    )
    df_agg.columns = df_agg.columns.str.lower()
    return df_agg
```

```python
Pluto_2009_agg=Pluto_agg(Pluto_2009)
Pluto_2010_agg=Pluto_agg(Pluto_2010)
Pluto_2011_agg=Pluto_agg(Pluto_2011)
Pluto_2012_agg=Pluto_agg(Pluto_2012)
Pluto_2013_agg=Pluto_agg(Pluto_2013)
Pluto_2014_agg=Pluto_agg(Pluto_2014)
Pluto_2015_agg=Pluto_agg(Pluto_2015)
Pluto_2016_agg=Pluto_agg(Pluto_2016)
Pluto_2017_agg=Pluto_agg(Pluto_2017)
```

```python
def Pluto_agg_2018_2020(df):

    df_agg=(df
                .groupby(['full_tract', 'year'])
                [['lotarea', 'bldgarea', 'comarea', 'resarea', 'officearea',
                    'retailarea', 'garagearea', 'strgearea', 'factryarea',
                'otherarea', 'numbldgs', 'numfloors', 'unitsres', 'unitstotal',
                'assessland', 'assesstot']]
                 .agg('sum')
                 .reset_index())
    return df_agg
```

```python
Pluto_2018_agg=Pluto_agg_2018_2020(Pluto_2018_MN)
Pluto_2019_agg=Pluto_agg_2018_2020(Pluto_2019_MN)
Pluto_2020_agg=Pluto_agg_2018_2020(Pluto_2020_MN)
```

```python
pluto_agg_MN_all_years = pd.concat([
    Pluto_2009_agg, Pluto_2010_agg, Pluto_2011_agg,
    Pluto_2012_agg, Pluto_2013_agg, Pluto_2014_agg,
    Pluto_2015_agg, Pluto_2016_agg, Pluto_2017_agg,
    Pluto_2018_agg, Pluto_2019_agg, Pluto_2020_agg
])
pluto_agg_MN_all_years = pluto_agg_MN_all_years.sort_values(by=['full_tract', 'year'])
pluto_agg_MN_all_years
```

```python

def convert_sqft_to_sqm(df, area_columns, suffix="_sqm"):
    sqft_to_sqm = 0.092903
    for col in area_columns:
        new_col = f"{col}{suffix}"
        df[new_col] = df[col] * sqft_to_sqm
    return df

```

```python
area_columns = ['lotarea', 'bldgarea', 'comarea', 'resarea', 'officearea',
                'retailarea', 'garagearea', 'strgearea', 'factryarea',
                'otherarea']


```

```python
pluto_agg_MN_all_years=convert_sqft_to_sqm(pluto_agg_MN_all_years, area_columns)
```

```python
# Specify the columns to plot 
sqm_cols = ['bldgarea_sqm', 'comarea_sqm', 'officearea_sqm']

# Filter for Hudson Yards
tract = '36061009900' # in Pluto there is no division for 36061009903, 36061009902, 36061009901. 
df_filtered = pluto_agg_MN_all_years[pluto_agg_MN_all_years['full_tract'] == tract]

# Aggregate by year
df_aggregated = df_filtered.groupby('year')[sqm_cols].sum().reset_index()


def millions(x, pos):
    return f'{x * 1e-6:.1f}M'

formatter = FuncFormatter(millions)


label_map = {
    'bldgarea_sqm': 'Total Built Area',
    'comarea_sqm': 'Commercial Area',
    'officearea_sqm': 'Office Space Area'
    
}

plt.figure(figsize=(10, 5))
for col in sqm_cols:
    plt.plot(df_aggregated['year'], df_aggregated[col], label=label_map[col])


# Vertical line and annotation

special_year=2015
plt.axvline(x=special_year, color="#C91441", linestyle='--', linewidth=2)
plt.text(special_year + 0.1, plt.ylim()[1] * 0.95, 'Subway opens (September 2015)', color='#800020')


plt.title('Panel A. Area Built in Hudson Yards', fontsize=14)
plt.xlabel('Year', fontsize=13)
plt.ylabel('Area (sq. meters)', fontsize=13)
plt.legend(fontsize=11,loc='upper left')
plt.grid(True)

# Make y-axis show millions
plt.gca().yaxis.set_major_formatter(formatter)

plt.tight_layout()
plt.show()
```
